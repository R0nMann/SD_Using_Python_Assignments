```python
def upper_case_lines():
  count=0
  f=open("/content/file.txt","r")
  for line in f:
    print(line)
    if line[0].isupper():
      count+=1
  return count
  f.close()
print("Number of lines: ",upper_case_lines())
```

```python
def display_two_char_words():
    f = open("/content/file.txt", "r")

    for line in f:
        words = line.split()
        print("All words: ",words)
        for w in words:
            if len(w) == 2:
                print(w)

    f.close()
print(display_two_char_words())
```

```python
def COUNT(word):
  freq=0
  f=open("/content/file.txt","r")
  for line in f:
    words = line.split()
    for w in words:
      if w==word:
        freq +=1
  f.close()
  print("Word:", word)
  print("Number of occurrences:", freq)
print(COUNT("school!"))
```

```python
def dispS():
  f=open("/content/file.txt","r")
  for line in f:
    if line[0]=='S':
      print(line)
  f.close()
f=open("/content/file.txt","r")
for x in f:
  print(x)
f.close()
print("------------------------------")
print(dispS())
```

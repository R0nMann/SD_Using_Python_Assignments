```python
# 1. Write a Python program to compute element-wise sum of given tuples
tuple1 = (1, 2, 3, 4, 5)
tuple2 = (10, 20, 30, 40, 50)
tuple3 = (100, 200, 300, 400, 500)

result = tuple(map(sum, zip(tuple1, tuple2, tuple3)))
print(f"Tuple 1 : {tuple1}")
print(f"Tuple 2 : {tuple2}")
print(f"Tuple 3 : {tuple3}")
print(f"Element-wise Sum : {result}")
```

```python
# 2. Write a python program that has a nested list to store toppers details. Edit the details and reprint the details.
toppers = [
    ["Alice",   "10-A", 98],
    ["Bob",     "10-B", 95],
    ["Charlie", "11-A", 97],
    ["Diana",   "11-B", 99],
]

print("\nOriginal Toppers List:")
print(f"{'Name':<12} {'Class':<8} {'Marks'}")
print("-" * 30)
for t in toppers:
    print(f"{t[0]:<12} {t[1]:<8} {t[2]}")

# Editing: update Bob's marks and Charlie's class
toppers[1][2] = 96          # Bob's marks corrected to 96
toppers[2][1] = "11-C"      # Charlie moved to 11-C

print("\nUpdated Toppers List (Bob marks→96, Charlie class→11-C):")
print(f"{'Name':<12} {'Class':<8} {'Marks'}")
print("-" * 30)
for t in toppers:
    print(f"{t[0]:<12} {t[1]:<8} {t[2]}")
```

```python
# 3. Write a Python program to find the repeated items of a tuple
sample = (1, 3, 5, 1, 7, 3, 9, 5, 5, 2)
print(f"Tuple : {sample}")

repeated = tuple(item for item in set(sample) if sample.count(item) > 1)
print(f"Repeated items : {repeated}")
```

```python
# 4. Write a Python program to Sort by Frequency of second element in Tuple List
data = [('a', 2), ('b', 3), ('c', 2), ('d', 5), ('e', 3), ('f', 2)]
print(f"Original : {data}")

# Count frequency of each second element
freq = {}
for item in data:
    freq[item[1]] = freq.get(item[1], 0) + 1

sorted_data = sorted(data, key=lambda x: freq[x[1]], reverse=True)
print(f"Sorted by 2nd-element frequency (desc) : {sorted_data}")
```

```python
# 5. Write a Python program to Sort lists in tuple
mixed = ([3, 1, 2], [9, 4, 7], [5, 8, 6])
print(f"Original tuple of lists : {mixed}")

sorted_tuple = tuple(sorted(lst) for lst in mixed)
print(f"After sorting each list : {sorted_tuple}")
```

```python
# 6. Write a Python program to Order Tuples using external List
tuples_list   = [('banana', 2), ('apple', 4), ('cherry', 1), ('date', 3)]
order_list    = ['apple', 'banana', 'cherry', 'date']   # desired order

print(f"Tuples     : {tuples_list}")
print(f"Order list : {order_list}")

ordered = sorted(tuples_list, key=lambda x: order_list.index(x[0]))
print(f"Ordered    : {ordered}")
```

```python
# 7. Write a Python program to Filter Tuples by Kth element from List
records   = [(1, 'a', 10), (2, 'b', 20), (3, 'a', 30), (4, 'c', 40), (5, 'b', 50)]
filter_val = 'a'          # keep tuples whose Kth element matches
K          = 1            # 0-indexed → second element

print(f"Records      : {records}")
print(f"Filter value : '{filter_val}'  |  K (index) : {K}")

filtered = [t for t in records if t[K] == filter_val]
print(f"Filtered     : {filtered}")
```

```python
# 8. Write a Python program to find a touple from a nested touple with respect to key element search which is the first element of all
# the nested touples .

nested = (
    (101, 'Alice',   'Math',    98),
    (102, 'Bob',     'Science', 95),
    (103, 'Charlie', 'English', 97),
    (104, 'Diana',   'Math',    99),
)
search_key = 103
print(f"Nested tuple (first element is key):")
for t in nested:
    print(f"  {t}")

found = next((t for t in nested if t[0] == search_key), None)
print(f"\nSearching for key = {search_key}")
print(f"Result : {found}" if found else "Result : Key not found")
```

```python
# 9. Write a Python program to Sort Tuples by their Maximum element
tuples_max = [(3, 1, 4), (1, 5, 9), (2, 6, 5), (8, 9, 7), (0, 2, 1)]
print(f"Original : {tuples_max}")

sorted_by_max = sorted(tuples_max, key=lambda x: max(x))
print(f"Sorted (ascending by max) : {sorted_by_max}")

sorted_by_max_desc = sorted(tuples_max, key=lambda x: max(x), reverse=True)
print(f"Sorted (descending by max): {sorted_by_max_desc}")
```
